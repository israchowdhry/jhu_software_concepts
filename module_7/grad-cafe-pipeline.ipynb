```python
"""Download GradCafe data from S3 and convert it from JSONL to JSON."""

from pathlib import Path

from src.s3_fetch import download_from_s3, jsonl_to_json

BUCKET_NAME = "grad-cafe-ic"
S3_OBJECT_KEY = "llm_extend_applicant_data.jsonl"
LOCAL_JSONL_PATH = "llm_extend_applicant_data_SM.jsonl"
LOCAL_JSON_PATH = "applicant_data_SM.json"


def main():
    """Download the source JSONL file and create the required JSON file."""
    local_jsonl = download_from_s3(
        bucket=BUCKET_NAME,
        key=S3_OBJECT_KEY,
        dest_path=LOCAL_JSONL_PATH,
    )

    local_json = jsonl_to_json(
        jsonl_path=str(local_jsonl),
        json_path=LOCAL_JSON_PATH,
    )

    print(f"Downloaded: {local_jsonl}")
    print(f"Created required JSON: {local_json}")
    print(f"Exists? {Path(LOCAL_JSON_PATH).exists()}")


main()
```
